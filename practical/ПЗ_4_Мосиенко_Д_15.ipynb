## Задача № 2

Рекурсивная функция для подсчета элементов списка

- Базовый случай: пустой список, элементов 0
- Шаг рекурсии: 1 (первый элемент) плюс количество элементов в остатке списка

````python
def count_items(lst):
    if not lst:                          # базовый случай
        return 0
    return 1 + count_items(lst[1:])      # один элемент + все остальные


fruits = ["яблоко", "груша", "слива", "вишня", "манго"]
print("Список:", fruits)
print("Количество элементов:", count_items(fruits))
print("Пустой список:", count_items([]))
````

## Задача № 3

Рекурсивная функция для поиска наибольшего числа в списке

- Базовый случай: в списке один элемент, он и есть максимум
- Шаг рекурсии: находим максимум среди остатка списка и сравниваем его с первым элементом

````python
def find_max(lst):
    if not lst:                          # пустой список, максимума нет
        return None
    if len(lst) == 1:                    # базовый случай
        return lst[0]
    rest_max = find_max(lst[1:])         # максимум в остатке списка
    return lst[0] if lst[0] > rest_max else rest_max


print("Массив:", nums)
print("Максимум:", find_max(nums))
print("Проверка через max():", max(nums))
print("Список с отрицательными:", find_max([-7, -3, -15, -1, -9]))
````

## Задача № 5

Быстрая сортировка и сортировка слиянием

- verbose=True - выводит шаги алгоритма
- pivot - опорный элемент в быстрой сортировке: "first", "last" или "random"
- reverse=True - сортировка по убыванию, False - по возрастанию

````python
def quick_sort(arr, pivot="last", reverse=False, verbose=False):
    if len(arr) <= 1:  # базовый случай
        return arr

    # индекс опорного элемента
    if pivot == "first":
        i = 0
    elif pivot == "last":
        i = len(arr) - 1
    else:
        i = random.randint(0, len(arr) - 1)

    p = arr[i]
    rest = arr[:i] + arr[i + 1:]  # без опорного

    # делим на левую и правую части
    if reverse:
        left = [x for x in rest if x > p]
        right = [x for x in rest if x <= p]
    else:
        left = [x for x in rest if x <= p]
        right = [x for x in rest if x > p]

    if verbose:
        print(arr, "опорный", p, "слева", left, "справа", right)

    return quick_sort(left, pivot, reverse, verbose) + [p] + quick_sort(right, pivot, reverse, verbose)


def merge_sort(arr, reverse=False, verbose=False):
    if len(arr) <= 1:  # базовый случай
        return arr

    # делим пополам
    mid = len(arr) // 2
    left = merge_sort(arr[:mid], reverse, verbose) # рекурсивно для левой и правой
    right = merge_sort(arr[mid:], reverse, verbose)

    if verbose:
        print("делим", arr, "на", arr[:mid], "и", arr[mid:], "-> сливаем", left, "и", right)

    # слияние двух частей
    result = []
    i = j = 0
    while i < len(left) and j < len(right): # поочередно сравнивает элементы из обоих частей
        if (left[i] <= right[j]) != reverse:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result += left[i:] + right[j:]  # остаток
    return result
````

### Тест 1. Массив из исходных данных

````python
print("Исходный массив:", nums)

print("\nБыстрая сортировка, опорный первый, по возрастанию")
print(quick_sort(nums, "first", False, True))

print("\nБыстрая сортировка, опорный последний, по убыванию")
print(quick_sort(nums, "last", True, True))

print("\nБыстрая сортировка, опорный случайный, по возрастанию")
print(quick_sort(nums, "random", False, True))

print("\nСортировка слиянием, по возрастанию")
print(merge_sort(nums, False, True))

print("\nСортировка слиянием, по убыванию")
print(merge_sort(nums, True))
````

### Тест 3. Массивы на 20, 30 и 50 элементов с повторами

Здесь без подробного вывода шагов, только результат и проверка через `sorted()`

````python
for n, low, high in ((20, 0, 100), (30, -100, 100), (50, -1000, 1000)):
    arr = [random.randint(low, high) for _ in range(n)]
    arr[-1] = arr[0]  # гарантируем повтор
    print("Исходный массив из", n, "элементов:", arr)
    print("Отсортированный:", merge_sort(arr))

    for rev in (False, True):
        ok1 = quick_sort(arr, "random", rev) == sorted(arr, reverse=rev)
        ok2 = merge_sort(arr, rev) == sorted(arr, reverse=rev)
        print("по убыванию:", rev, "| быстрая верно:", ok1, "| слиянием верно:", ok2)
    print()
````

## Задача № 6

Дан отсортированный массив с положительными и отрицательными элементами, нужно отсортировать его по модулю

**Идея:** массив уже упорядочен по значению, поэтому самый большой по модулю элемент стоит либо слева (самый отрицательный), либо справа (самый большой положительный)

- ставим два указателя на левый и правый конец
- сравниваем модули элементов под указателями, больший кладем в результат с конца и двигаем этот указатель
- повторяем, пока не заполним весь результат

**Сложность:** один проход, O(n) по времени и O(n) по памяти, это лучше, чем сортировать заново за O(n log n)

````python
def sort_by_abs(arr):
    n = len(arr)
    result = [0] * n
    left, right = 0, n - 1

    # заполняем результат с конца, от большего модуля к меньшему
    for pos in range(n - 1, -1, -1):
        if abs(arr[left]) > abs(arr[right]):
            result[pos] = arr[left]
            left += 1
        else:
            result[pos] = arr[right]
            right -= 1
    return result


tests = [
    [-10, -7, -3, 1, 4, 6, 9],
    [-9, -8, -2, 0, 3, 5, 7],
    [-5, -5, 5, 5],
    [-4, -3, -2, -1],
    [1, 2, 3, 4],
    [-6, 6],
    [],
]

for t in tests:
    res = sort_by_abs(t)
    print(t, "-", res, "-", res == sorted(t, key=abs))
````
