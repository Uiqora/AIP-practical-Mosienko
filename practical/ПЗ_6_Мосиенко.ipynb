```python
class Node:
    # узел связного списка: значение и ссылка на следующий узел
    def __init__(self, value, link_next=None):
        self.value = value
        self.link_next = link_next
```

## Задача № 1

Реализовать однонаправленный связный список в виде класса, методы: чтение всех элементов, вставка в конец, вставка в начало, удаление с конца, удаление с начала, поиск элемента, вставка после указанного элемента, удаление указанного элемента

В списке хранится только ссылка на первый узел (head), остальные узлы достаются переходом по ссылкам link_next  
Вставка и удаление в начале выполняются за O(1), остальные операции требуют прохода по списку, O(n)

- Чтение: идем от head до узла, у которого ссылка None
- Вставка в конец: доходим до последнего узла и привязываем к нему новый
- Вставка в начало: новый узел ссылается на старый head и сам становится head
- Удаление с конца: доходим до предпоследнего узла и обнуляем его ссылку
- Удаление с начала: head переходит на следующий узел
- Поиск: проходим по списку и сравниваем значения, возвращаем узел или None
- Вставка после элемента: находим узел, новый узел ссылается на его следующий, сам узел ссылается на новый
- Удаление элемента: находим предыдущий узел и перекидываем его ссылку через удаляемый

```python
class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def read_all(self):
        # чтение всех элементов, возвращаем обычный список
        res = []
        node = self.head
        while node is not None:
            res.append(node.value)
            node = node.link_next
        return res

    def append_to_tail(self, value):
        node_end = Node(value)
        if self.head is None:  # список пустой, новый узел становится первым
            self.head = node_end
            return
        node = self.head
        while node.link_next is not None:  # идем до последнего узла
            node = node.link_next
        node.link_next = node_end

    def append_to_head(self, value):
        # новый узел ссылается на старый head, потом сам становится head
        self.head = Node(value, self.head)

    def remove_from_tail(self):
        if self.head is None:
            return None
        if self.head.link_next is None:  # в списке один узел
            value = self.head.value
            self.head = None
            return value
        node = self.head
        while node.link_next.link_next is not None:  # останавливаемся на предпоследнем
            node = node.link_next
        value = node.link_next.value
        node.link_next = None  # последний узел отсоединен, на него больше нет ссылок
        return value

    def remove_from_head(self):
        if self.head is None:
            return None
        value = self.head.value
        self.head = self.head.link_next
        return value

    def find(self, value):
        node = self.head
        while node is not None:
            if node.value == value:
                return node  # возвращаем сам узел, а не значение
            node = node.link_next
        return None

    def insert_after(self, target, value):
        # вставка после первого узла со значением target, True если получилось
        node = self.find(target)
        if node is None:
            return False
        # новый узел сначала цепляется к следующему, иначе хвост списка потеряется
        node.link_next = Node(value, node.link_next)
        return True

    def remove(self, value):
        # удаление первого узла с указанным значением, True если получилось
        if self.head is None:
            return False
        if self.head.value == value:  # удаляем первый узел
            self.head = self.head.link_next
            return True
        node = self.head
        while node.link_next is not None:
            if node.link_next.value == value:  # смотрим на следующий узел, чтобы остаться перед удаляемым
                node.link_next = node.link_next.link_next  # перекидываем ссылку через удаляемый
                return True
            node = node.link_next
        return False
```

```python
lst = SinglyLinkedList()

for x in (10, 11, 12):
    lst.append_to_tail(x)
print('Вставка в конец 10, 11, 12:', lst.read_all())

lst.append_to_head(9)
print('Вставка в начало 9:', lst.read_all())

print('Поиск 11:', lst.find(11).value if lst.find(11) else None)  # проверка, чтобы не взять .value у None
print('Поиск 99:', lst.find(99))

lst.insert_after(11, 55)
print('Вставка 55 после 11:', lst.read_all())
print('Вставка после несуществующего 99:', lst.insert_after(99, 1))

lst.remove(55)
print('Удаление 55:', lst.read_all())
print('Удаление несуществующего 99:', lst.remove(99))

print('Удаление с конца:', lst.remove_from_tail(), lst.read_all())
print('Удаление с начала:', lst.remove_from_head(), lst.read_all())

# опустошаем список и проверяем граничные случаи
lst.remove_from_tail()
lst.remove_from_tail()
print('Пустой список:', lst.read_all(), lst.is_empty())
print('Удаление из пустого:', lst.remove_from_tail(), lst.remove_from_head())
```

## Задача № 2

Реализовать двунаправленный связный список в виде класса, методы: чтение всех элементов, вставка в конец, удаление с конца

Каждый узел хранит две ссылки: на следующий (link_next) и на предыдущий (link_prev)  
В списке храним ссылки на первый (head) и последний (tail) узлы, поэтому вставка и удаление в конце выполняются за O(1), проходить весь список не нужно

- Вставка в конец: новый узел ссылается назад на tail, старый tail ссылается вперед на новый, tail переходит на новый узел
- Удаление с конца: tail переходит на предыдущий узел, у него обнуляется ссылка вперед

```python
class DNode:
    # узел двунаправленного списка
    def __init__(self, value, link_next=None, link_prev=None):
        self.value = value
        self.link_next = link_next
        self.link_prev = link_prev  # ссылка назад, позволяет идти от конца к началу


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def read_all(self):
        res = []
        node = self.head
        while node is not None:
            res.append(node.value)
            node = node.link_next
        return res

    def read_all_reverse(self):
        # для проверки обратных ссылок идем от tail к head
        res = []
        node = self.tail
        while node is not None:
            res.append(node.value)
            node = node.link_prev
        return res

    def append_to_tail(self, value):
        node_end = DNode(value, None, self.tail)  # новый узел смотрит назад на старый tail
        if self.tail is None:  # список пустой
            self.head = node_end
        else:
            self.tail.link_next = node_end  # старый tail смотрит вперед на новый
        self.tail = node_end

    def remove_from_tail(self):
        if self.tail is None:
            return None
        value = self.tail.value
        self.tail = self.tail.link_prev
        if self.tail is None:  # был один узел
            self.head = None
        else:
            self.tail.link_next = None  # иначе ссылка вела бы на удаленный узел
        return value
```

```python
dl = DoublyLinkedList()

for x in (1, 2, 3, 4, 5):
    dl.append_to_tail(x)
print('Вставка в конец 1..5')
print('Вперед:', dl.read_all())
print('Назад: ', dl.read_all_reverse())

print('Удаление с конца:', dl.remove_from_tail())
print('Удаление с конца:', dl.remove_from_tail())
print('Вперед:', dl.read_all())
print('Назад: ', dl.read_all_reverse())

# граничные случаи
for i in range(3):
    dl.remove_from_tail()
print('Пустой список:', dl.read_all(), dl.head, dl.tail)
print('Удаление из пустого:', dl.remove_from_tail())
```

## Задача № 3

Реализовать циклический связный список в виде класса, методы: чтение всех элементов, вставка в конец, удаление с конца

В циклическом списке последний узел ссылается не на None, а на первый  
Храним ссылку только на последний узел (tail), первый узел всегда доступен как tail.link_next  
Из-за цикла нельзя ждать None, поэтому при обходе останавливаемся, когда вернулись к первому узлу

- Вставка в конец: новый узел ссылается на первый, старый tail ссылается на новый, tail переходит на новый узел
- Удаление с конца: доходим до предпоследнего узла, он становится tail и замыкается на первый узел

```python
class CircularLinkedList:
    def __init__(self):
        self.tail = None  # храним только последний узел, первый это tail.link_next

    def read_all(self):
        res = []
        if self.tail is None:
            return res
        first = self.tail.link_next
        node = first
        while True:
            res.append(node.value)
            node = node.link_next
            if node is first:  # вернулись к началу, круг пройден
                break
        return res

    def append_to_tail(self, value):
        node_end = Node(value)
        if self.tail is None:  # единственный узел ссылается сам на себя
            node_end.link_next = node_end
        else:
            node_end.link_next = self.tail.link_next  # новый узел замыкается на первый
            self.tail.link_next = node_end  # старый последний ссылается на новый
        self.tail = node_end

    def remove_from_tail(self):
        if self.tail is None:
            return None
        value = self.tail.value
        if self.tail.link_next is self.tail:  # в списке один узел
            self.tail = None
            return value
        node = self.tail.link_next
        while node.link_next is not self.tail:  # None в кольце нет, ищем узел перед tail
            node = node.link_next
        node.link_next = self.tail.link_next  # замыкаем на первый
        self.tail = node
        return value
```

```python
cl = CircularLinkedList()

for x in (7, 8, 9, 10):
    cl.append_to_tail(x)
print('Вставка в конец 7..10:', cl.read_all())
print('Последний ссылается на первый:', cl.tail.link_next.value == cl.read_all()[0])

print('Удаление с конца:', cl.remove_from_tail(), cl.read_all())
print('Удаление с конца:', cl.remove_from_tail(), cl.read_all())
print('Последний ссылается на первый:', cl.tail.link_next.value == cl.read_all()[0])

# граничные случаи
cl.remove_from_tail()
print('Один узел:', cl.read_all())
cl.remove_from_tail()
print('Пустой список:', cl.read_all(), cl.tail)
print('Удаление из пустого:', cl.remove_from_tail())
```

## Задача № 4

На основе собственного класса реализовать очередь через связный список и решить задачи с ее помощью

Очередь работает по принципу первым пришел, первым ушел  
Основа это однонаправленный связный список из задачи № 1 (узлы Node), в очереди храним ссылки на начало (head) и конец (tail)

- enqueue: добавляем узел после tail, O(1)
- dequeue: забираем значение из head и сдвигаем head, O(1)
- ссылка на tail позволяет не проходить по всему списку при добавлении

```python
class LinkedQueue:
    def __init__(self, items=()):
        self.head = None
        self.tail = None
        self.size = 0
        for x in items:
            self.enqueue(x)

    def is_empty(self):
        return self.head is None

    def enqueue(self, value):
        # добавление в конец очереди
        node = Node(value)
        if self.tail is None:  # очередь была пуста
            self.head = node
        else:
            self.tail.link_next = node  # tail позволяет добавлять без прохода по списку
        self.tail = node
        self.size += 1

    def dequeue(self):
        # извлечение из начала очереди
        if self.head is None:
            return None
        value = self.head.value
        self.head = self.head.link_next
        if self.head is None:  # очередь опустела, tail не должен указывать на удаленный узел
            self.tail = None
        self.size -= 1
        return value

    def to_list(self):
        # просмотр очереди без извлечения (для вывода)
        res = []
        node = self.head
        while node is not None:
            res.append(node.value)
            node = node.link_next
        return res
```

### Задача 4.1

Дан набор из N чисел (N > 10), создать очередь из этих чисел в том же порядке (первое число в начале, последнее в конце), извлечь все элементы, вывести их и количество извлеченных

Заполняем очередь через enqueue, потом вызываем dequeue пока очередь не пуста и считаем элементы

```python
data = [11, 5, 8, 20, 3, 14, 9, 6, 17, 2, 13, 1]
print('N =', len(data), 'числа:', data)

q = LinkedQueue(data)
print('Очередь:', q.to_list())

res = []
while not q.is_empty():
    res.append(q.dequeue())

print('Извлечено:', res)
print('Количество извлеченных элементов N =', len(res))
```

### Задача 4.2

Даны две очереди, переместить все элементы первой очереди (от начала к концу) в конец второй очереди

Пока первая очередь не пуста, извлекаем элемент из ее начала и добавляем в конец второй  
Порядок элементов сохраняется

```python
def move_all(q1, q2):
    while not q1.is_empty():
        q2.enqueue(q1.dequeue())  # достаем из начала первой и сразу кладем в конец второй


q1 = LinkedQueue([1, 2, 3, 4])
q2 = LinkedQueue([10, 20])
print('До:   ', q1.to_list(), q2.to_list())

move_all(q1, q2)
print('После:', q1.to_list(), q2.to_list())
```

### Задача 4.3

Дано число N (> 0) и две непустые очереди, создать функцию для перемещения N начальных элементов первой очереди в конец второй, если в первой очереди меньше N элементов, переместить все

Повторяем N раз: извлекаем элемент из первой очереди и добавляем во вторую  
Если первая очередь закончилась раньше, цикл прерываем

```python
def move_n(q1, q2, n):
    for i in range(n):
        if q1.is_empty():  # элементов меньше N, перенесли все
            break
        q2.enqueue(q1.dequeue())


a = [3, 6, 9, 12, 15, 18]
b = [1, 2]

for n in (4, 10):
    q1, q2 = LinkedQueue(a), LinkedQueue(b)
    move_n(q1, q2, n)
    print('N =', n)
    print('Первая: ', q1.to_list())
    print('Вторая: ', q2.to_list())
```
