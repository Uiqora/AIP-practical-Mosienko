## Задача 2


```python
list1 = [10, 15, 20, 25, 500]

for element in list1:
    print(element, end='\t')

for i in range(5):
    print(list1[i])
```

Программа состоит из двух последовательных циклов.

Первый цикл проходит по всем элементам списка list1, т.е. выполняется n раз, где n - длина списка.
Его сложность O(n).

Второй цикл формально идет по range(5), но 5 - это как раз длина того же списка (если бы список был длиннее,
пришлось бы поменять и границу цикла), поэтому по смыслу он тоже линейный, O(n).

По закону сложения итоговая сложность равна сумме сложностей двух действий, а из суммы слагаемых
оставляют только самое "медленное", отбрасывая константы:

O(n) + O(n) = O(n)

Итоговая сложность: O(n).

## Задача 3


```python
list1 = [10, 15, 20, 25, 500]
n = len(list1)
for i in range(n):
    el1 = [list1[i]]
    for j in range(n):
        el2 = [list1[j]]
        if el1 != el2:
            print(el1, '--', el2)
```

Внешний цикл проходит n раз, и для каждого его шага внутренний цикл тоже проходит n раз.
Действия внутри внутреннего цикла (сравнение, print) выполняются за постоянное время, на сложность не влияют.

По закону умножения (вложенные действия перемножаются):

O(n) * O(n) = O(n^2)

Итоговая сложность: O(n^2).

## Задача 4

Дан список: [6, 12, 4, 3, 8]. Отсортировать его по возрастанию сортировкой выбором.

### 1. Описание

Работаем с одним массивом, на каждом шаге ищем минимум в необработанной части и ставим его на очередное место.

Исходный массив: [6, 12, 4, 3, 8]

Шаг 1. Минимум среди [6, 12, 4, 3, 8] - это 3 (индекс 3). Меняем местами элементы с индексами 0 и 3.
Получаем [3, 12, 4, 6, 8].

Шаг 2. Минимум среди оставшейся части [12, 4, 6, 8] (индексы 1-4) - это 4 (индекс 2). Меняем местами
элементы с индексами 1 и 2. Получаем [3, 4, 12, 6, 8].

Шаг 3. Минимум среди [12, 6, 8] (индексы 2-4) - это 6 (индекс 3). Меняем местами элементы с индексами 2 и 3.
Получаем [3, 4, 6, 12, 8].

Шаг 4. Минимум среди [12, 8] (индексы 3-4) - это 8 (индекс 4). Меняем местами элементы с индексами 3 и 4.
Получаем [3, 4, 6, 8, 12].

Массив отсортирован: [3, 4, 6, 8, 12].

### 2. Функция с циклом for

```python
def selection_sort_for(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
    return a


list1 = [6, 12, 4, 3, 8]
print(selection_sort_for(list1))
```

### 3. Функция с циклом while

```python
def selection_sort_while(arr):
    a = arr.copy()
    n = len(a)
    i = 0
    while i < n:
        min_idx = i
        j = i + 1
        while j < n:
            if a[j] < a[min_idx]:
                min_idx = j
            j += 1
        a[i], a[min_idx] = a[min_idx], a[i]
        i += 1
    return a


print(selection_sort_while(list1))
```

### 4. Сортировка выбором через 2 массива

```python
def selection_sort_two_arrays(arr):
    source = arr.copy()
    result = []
    while source:
        min_idx = 0
        for j in range(1, len(source)):
            if source[j] < source[min_idx]:
                min_idx = j
        result.append(source.pop(min_idx))
    return result


print(selection_sort_two_arrays(list1))
```

### 5. Сортировка выбором через 1 массив (еще один вариант)

```python
def selection_sort_one_array(arr):
    a = arr.copy()
    n = len(a)
    start = 0
    while start < n:
        min_idx = start
        for j in range(start, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[start], a[min_idx] = a[min_idx], a[start]
        start += 1
    return a


print(selection_sort_one_array(list1))
```

Все четыре варианта дают одинаковый результат. Для дальнейшей работы (задача 6) буду использовать
функцию selection_sort_for - она компактнее остальных.

## Задача 5

Отсортировать список из задачи 4 по возрастанию сортировкой пузырьком.

### 1. Словесное описание

Суть - сравнивать соседние элементы и menять их местами, если левый больше правого. За каждый проход
самый большой из необработанных элементов "всплывает" в конец.

Исходный массив: [6, 12, 4, 3, 8]

Проход 1 (4 сравнения):
6 и 12 - не меняем
12 и 4 - меняем -> [6, 4, 12, 3, 8]
12 и 3 - меняем -> [6, 4, 3, 12, 8]
12 и 8 - меняем -> [6, 4, 3, 8, 12]
После прохода: [6, 4, 3, 8, 12], число 12 встало на свое место.

Проход 2 (3 сравнения):
6 и 4 - меняем -> [4, 6, 3, 8, 12]
6 и 3 - меняем -> [4, 3, 6, 8, 12]
6 и 8 - не меняем
После прохода: [4, 3, 6, 8, 12]

Проход 3 (2 сравнения):
4 и 3 - меняем -> [3, 4, 6, 8, 12]
4 и 6 - не меняем
После прохода: [3, 4, 6, 8, 12]

Проход 4 (1 сравнение):
3 и 4 - не меняем
После прохода: [3, 4, 6, 8, 12]

Массив отсортирован, дальше проходов не требуется: [3, 4, 6, 8, 12].

### 2. Функция с циклом for

```python
def bubble_sort_for(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a


list1 = [6, 12, 4, 3, 8]
print(bubble_sort_for(list1))
```

### 3. Функция с циклом while

```python
def bubble_sort_while(arr):
    a = arr.copy()
    n = len(a)
    i = 0
    while i < n - 1:
        j = 0
        while j < n - 1 - i:
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
            j += 1
        i += 1
    return a


print(bubble_sort_while(list1))
```

Обе версии сортируют список одинаково. Дальше (задача 6) буду использовать bubble_sort_for.

## Задача 6

Добавить в функции selection_sort_for и bubble_sort_for параметр verbose. Если verbose=True, в консоль
должны выводиться: исходный массив, каждый шаг алгоритма, массив после каждого шага, а в конце - сколько
шагов потребовалось и отсортированный массив.

За "шаг" беру одну итерацию внутреннего цикла (одно сравнение, возможно со сменой мест) - так удобнее
проверять правильность работы алгоритма.

```python
def selection_sort(arr, verbose=False):
    a = arr.copy()
    n = len(a)
    step = 0
    if verbose:
        print('Исходный массив:', a)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            step += 1
            if a[j] < a[min_idx]:
                min_idx = j
            if verbose:
                print(f'Шаг {step}: сравниваем {a[j]} и {a[min_idx]}, массив пока не меняется -> {a}')
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
        if verbose:
            print(f'После обработки позиции {i}: {a}')
    if verbose:
        print('Всего шагов (сравнений):', step)
        print('Отсортированный массив:', a)
    return a


def bubble_sort(arr, verbose=False):
    a = arr.copy()
    n = len(a)
    step = 0
    if verbose:
        print('Исходный массив:', a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            step += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
            if verbose:
                print(f'Шаг {step}: {a}')
    if verbose:
        print('Всего шагов:', step)
        print('Отсортированный массив:', a)
    return a
```

Проверка на списке из задачи 4/5:

```python
list1 = [6, 12, 4, 3, 8]
print('--- selection_sort ---')
selection_sort(list1, verbose=True)
print()
print('--- bubble_sort ---')
bubble_sort(list1, verbose=True)
```

## Задача 7

Добавить в обе функции возможность сортировки по убыванию (аргумент reverse: False - по возрастанию,
True - по убыванию).

```python
def selection_sort(arr, verbose=False, reverse=False):
    a = arr.copy()
    n = len(a)
    step = 0
    if verbose:
        print('Исходный массив:', a)
    for i in range(n):
        target_idx = i
        for j in range(i + 1, n):
            step += 1
            if reverse:
                better = a[j] > a[target_idx]
            else:
                better = a[j] < a[target_idx]
            if better:
                target_idx = j
            if verbose:
                print(f'Шаг {step}: сравниваем {a[j]} и {a[target_idx]}')
        if target_idx != i:
            a[i], a[target_idx] = a[target_idx], a[i]
        if verbose:
            print(f'После обработки позиции {i}: {a}')
    if verbose:
        print('Всего шагов:', step)
        print('Отсортированный массив:', a)
    return a


def bubble_sort(arr, verbose=False, reverse=False):
    a = arr.copy()
    n = len(a)
    step = 0
    if verbose:
        print('Исходный массив:', a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            step += 1
            if reverse:
                need_swap = a[j] < a[j + 1]
            else:
                need_swap = a[j] > a[j + 1]
            if need_swap:
                a[j], a[j + 1] = a[j + 1], a[j]
            if verbose:
                print(f'Шаг {step}: {a}')
        i += 0
    if verbose:
        print('Всего шагов:', step)
        print('Отсортированный массив:', a)
    return a
```

```python
list1 = [6, 12, 4, 3, 8]
print('По возрастанию:')
print('selection_sort:', selection_sort(list1))
print('bubble_sort:', bubble_sort(list1))
print()
print('По убыванию:')
print('selection_sort:', selection_sort(list1, reverse=True))
print('bubble_sort:', bubble_sort(list1, reverse=True))
```

## Задача 9

Есть отсортированный массив L1 из 50 целых чисел в диапазоне от 0 до 100. Нужно найти элемент со
значением 10, при этом итоговая сложность должна быть минимальной.

Массив уже отсортирован, поэтому сортировать его заново не нужно (это было бы лишней и весьма дорогой
работой). Раз массив отсортирован - можно сразу применить бинарный поиск.

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    steps = 0
    while low <= high:
        steps += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, steps
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, steps
```

```python
np.random.seed(1)
L1 = sorted(int(x) for x in np.random.randint(0, 100, 49))
L1.append(10)
L1.sort()

print('L1:', L1)

idx, steps = binary_search(L1, 10)
print('Индекс найденного элемента:', idx)
print('Число шагов бинарного поиска:', steps)
```

Сортировка тут не нужна (массив и так отсортирован), значит на сложность влияет только сам поиск.
Бинарный поиск на каждом шаге отбрасывает половину массива, поэтому

Итоговая сложность: O(log n).

## Задача 10

Есть неотсортированный массив L1 из 50 целых чисел в диапазоне от 0 до 100. Нужно найти элемент со
значением 10, сложность должна быть минимальной.

Массив не отсортирован. Можно пойти двумя путями:

1) сначала отсортировать массив (сортировкой выбором или пузырьком, это O(n^2)), а потом применить
бинарный поиск (O(log n)) - итог O(n^2) + O(log n) = O(n^2);

2) сразу применить линейный поиск по неотсортированному массиву - O(n).

Так как задача - найти элемент один раз, а не выполнить много запросов подряд, сортировать массив
невыгодно: O(n^2) заведомо хуже, чем O(n). Поэтому эффективнее сразу использовать линейный поиск.

```python
def linear_search(arr, target):
    steps = 0
    for i in range(len(arr)):
        steps += 1
        if arr[i] == target:
            return i, steps
    return -1, steps
```

```python
np.random.seed(2)
L1 = [int(x) for x in np.random.randint(0, 100, 49)]
L1.append(10)
np.random.shuffle(L1)

print('L1:', L1)

idx, steps = linear_search(L1, 10)
print('Индекс найденного элемента:', idx)
print('Число шагов линейного поиска:', steps)
```

Итоговая сложность выбранного решения: O(n).

## Задача 11

Есть неотсортированный массив L1 из 50 целых чисел (0-100). Нужно: вывести первый элемент, найти
минимум, найти максимум, найти элемент со значением 10. Сложность должна быть минимальной.

Раз операций несколько и часть из них (min/max/search) все равно требует хотя бы одного прохода по
массиву, сравним два варианта:

1) без сортировки: первый элемент O(1), минимум - линейный проход O(n), максимум - линейный проход
O(n), поиск значения 10 - линейный проход O(n). По закону сложения итог O(1) + O(n) + O(n) + O(n) = O(n)
(константы отбрасываются, остается самое медленное слагаемое);

2) с сортировкой (доступны только O(n^2) варианты - выбором и пузырьком): сортировка O(n^2), после нее
первый элемент O(1), минимум - это первый элемент отсортированного массива O(1), максимум - последний
элемент O(1), поиск - бинарный поиск O(log n). Итог O(n^2) + O(1) + O(1) + O(1) + O(log n) = O(n^2).

O(n) меньше, чем O(n^2), поэтому сортировать массив ради этих операций невыгодно - эффективнее выполнить
все четыре действия напрямую, без сортировки.

```python
def find_min(arr):
    m = arr[0]
    for x in arr[1:]:
        if x < m:
            m = x
    return m


def find_max(arr):
    m = arr[0]
    for x in arr[1:]:
        if x > m:
            m = x
    return m
```

```python
np.random.seed(3)
L1 = [int(x) for x in np.random.randint(0, 100, 49)]
L1.append(10)
np.random.shuffle(L1)

print('L1:', L1)

first_el = L1[0]
min_val = find_min(L1)
max_val = find_max(L1)
idx, steps = linear_search(L1, 10)

print('Первый элемент:', first_el)
print('Минимум:', min_val)
print('Максимум:', max_val)
print('Индекс элемента 10:', idx)
```

Итоговая сложность выбранного решения (без сортировки): O(n).
