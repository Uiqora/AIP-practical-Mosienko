## Задача № 1

Дана функция funct(n), которая на каждом шаге берет последнюю цифру числа n (остаток от деления на 10), а само число уменьшает в 10 раз через целочисленное деление. Рекурсия заканчивается, когда число становится меньше 10, то есть остается одна цифра. Все взятые цифры складываются между собой. По сути функция считает сумму цифр числа.


```python
def funct(n):
    if n < 10:
        return n
    else:
        return n % 10 + funct(n // 10)

print(funct(230))
```

Стек вызовов для funct(230):


| № п/п | переменные | значения переменных | возвращаемая команда | возвращаемое значение |
|---|---|---|---|---|
| 3 | n | 2 | return n | 2 |
| 2 | n | 23 | n % 10 + funct(n // 10) | 3 + funct(2) = 5 |
| 1 | n | 230 | n % 10 + funct(n // 10) | 0 + funct(23) = 5 |

Итог: 230 -> 0 + 23 -> 23 -> 3 + 2 -> 2. Сумма цифр 0 + 3 + 2 = 5, что и выводится на экран.

## Задача № 2

Дана функция funct(k, n): k увеличивается на 1 на каждом рекурсивном шаге, пока не станет равно n (или больше). На каждом уровне текущее k добавляется в строку, а дальше идет вызов для k+1. Когда k достигает n, рекурсия останавливается и возвращается само n. Получается, что функция строит строку из чисел от k до n включительно, разделенных пробелом.

```python
def funct(k, n):
    if k < n:
        return str(k) + " " + str(funct(k + 1, n))
    else:
        return n

print(funct(4, 12))
```

| 8 | k, n | 11, 12 | str(k) + " " + str(funct(k+1, n)) | "11 12" |
| 7 | k, n | 10, 12 | str(k) + " " + str(funct(k+1, n)) | "10 11 12" |
| ... | ... | ... | ... | ... |

## Задача № 3

Нужно написать рекурсивную функцию, которая:

а) считает сумму всех чисел в списке;

б) считает сумму только положительных чисел, при этом в списке обязательно должно быть отрицательное число.

### а) Сумма всех чисел списка

```python
def sum_list(arr):
    # базовый случай - список закончился
    if not arr:
        return 0
    # рекурсивный случай - берем первый элемент и прибавляем сумму остальных
    return arr[0] + sum_list(arr[1:])


test_a = [10, 5, 5, 20]
print(sum_list(test_a))
```

### б) Сумма положительных чисел

По условию в списке обязательно должно быть хотя бы одно отрицательное значение. Логично использовать его как признак конца полезных данных: как только встречается отрицательное число, рекурсия останавливается, а до этого момента элементы складываются.

```python
def sum_positive(arr):
    # базовый случай — список закончился
    if not arr:
        return 0

    # если число отрицательное (или ноль) — пропускаем его и идем дальше
    if arr[0] <= 0:
        return sum_positive(arr[1:])

    # рекурсивный случай — элемент положительный, прибавляем и идем дальше
    return arr[0] + sum_positive(arr[1:])


test_b = [4, 7, 2, 9, -1, 100, 50]
print(sum_positive(test_b))
```

## Задача № 4

Нужно посчитать количество нечетных элементов массива двумя способами: через цикл и через рекурсию, а также расписать стек вызовов рекурсивного варианта.

### Через цикл

```python
def count_odd_loop(arr):
    count = 0
    for el in arr:
        if el % 2 != 0:
            count += 1
    return count


nums = [1, 2, 3, 4, 5, 6, 7]
print(count_odd_loop(nums))
```

### Через рекурсию

```python
def count_odd_recursive(arr):
    # базовый случай - массив пуст
    if not arr:
        return 0
    # если первый элемент нечетный, прибавляем 1 и идем дальше по остатку
    if arr[0] % 2 != 0:
        return 1 + count_odd_recursive(arr[1:])
    # если четный, просто идем дальше без прибавления
    else:
        return count_odd_recursive(arr[1:])


print(count_odd_recursive(nums))
```

Стек вызовов для count_odd_recursive([1, 2, 3, 4, 5, 6, 7]):


| № п/п | переменные | значения переменных | возвращаемая команда | возвращаемое значение |
|---|---|---|---|---|
| 8 | arr | [] | return 0 | 0 |
| 7 | arr | [7] | 1 + count_odd_recursive(arr[1:]) | 1 + 0 = 1 |
| 6 | arr | [6, 7] | count_odd_recursive(arr[1:]) | 1 |
| 5 | arr | [5, 6, 7] | 1 + count_odd_recursive(arr[1:]) | 1 + 1 = 2 |
| 4 | arr | [4, 5, 6, 7] | count_odd_recursive(arr[1:]) | 2 |
| 3 | arr | [3, 4, 5, 6, 7] | 1 + count_odd_recursive(arr[1:]) | 1 + 2 = 3 |
| 2 | arr | [2, 3, 4, 5, 6, 7] | count_odd_recursive(arr[1:]) | 3 |
| 1 | arr | [1, 2, 3, 4, 5, 6, 7] | 1 + count_odd_recursive(arr[1:]) | 1 + 3 = 4 |

Итог - 4 нечетных элемента (1, 3, 5, 7), оба способа дают одинаковый результат.




## Задача № 5

Даны последовательности скобок:

а) {}[()(( ))]

б) ()]()()[]

Нужно определить, является ли последовательность правильной (корректной), то есть каждой открывающей скобке соответствует своя закрывающая скобка того же типа, и порядок закрытия соблюдается.

### 1. Описание алгоритма (словами)

Идея - использовать стек:

1. Идем по строке слева направо, символ за символом.
2. Если встретили открывающую скобку ( (, [, { ), кладем ее в стек.
3. Если встретили закрывающую скобку, смотрим на верхний элемент стека:
   - если стек пуст - последовательность неправильная (закрывать нечего);
   - если верхний элемент стека - открывающая скобка того же типа, снимаем ее со стека;
   - если типы не совпадают - последовательность неправильная.
4. После прохода всей строки последовательность правильная только тогда, когда стек оказался пустым (все открытые скобки были закрыты).

Скобка снизу стека закрывается последней - это как раз принцип LIFO, о котором говорится в теории.

### 2. Реализация на Python

```python
def is_valid_brackets(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}

    for ch in s:
        if ch in '([{':
            stack.append(ch)
        elif ch in ')]}':
            if not stack:
                return False
            top = stack.pop()
            if top != pairs[ch]:
                return False
    # если стек не пустой, значит остались незакрытые скобки
    return len(stack) == 0


a = "{}[()(())]"
b = "()]()()[]"

print(a, "->", is_valid_brackets(a))
print(b, "->", is_valid_brackets(b))
```

Проверка вручную:

- Последовательность а) `{}[()(())]` - каждая скобка закрывается своей парой в правильном порядке, стек в конце пустой, значит последовательность правильная.
- Последовательность б) `()]()()[]` - третий символ `]` пытается закрыть скобку `(`, типы не совпадают, поэтому последовательность неправильная.


## Задача № 6

Дана строка из круглых скобок. Нужно найти минимальное количество символов, которые необходимо **удалить**, чтобы оставшаяся строка стала правильной скобочной последовательностью.

Идея простая: пробегаем по строке и считаем стеком открывающие скобки. Если встречаем закрывающую скобку, а стек пуст (открывать нечего) - эта закрывающая скобка лишняя, ее нужно удалить. В конце, если в стеке остались открывающие скобки, каждую из них тоже придется удалить.

### Итеративное решение

```python
def min_deletions_iterative(s):
    stack = []          # тут хранится количество открывающих скобок
    to_delete = 0        # количество лишних закрывающих скобок

    for ch in s:
        if ch == '(':
            stack.append(ch)
        elif ch == ')':
            if stack:
                stack.pop()   # нашлась пара, скобка не лишняя
            else:
                to_delete += 1  # закрывать нечего, скобка лишняя

    # все оставшиеся в стеке открывающие скобки тоже придется удалить
    to_delete += len(stack)
    return to_delete


print(min_deletions_iterative("()())("))
print(min_deletions_iterative(")()("))
print(min_deletions_iterative("((()"))
```

### Рекурсивное решение

```python
def min_deletions_recursive(s, index=0, open_count=0):
    # базовый случай - строка закончилась, удаляем все незакрытые открывающие скобки
    if index == len(s):
        return open_count

    ch = s[index]

    if ch == '(':
        return min_deletions_recursive(s, index + 1, open_count + 1)
    else:  # ch == ')'
        if open_count > 0:
            # нашлась пара для одной из открытых скобок
            return min_deletions_recursive(s, index + 1, open_count - 1)
        else:
            # закрывающая скобка лишняя, добавляем 1 к ответу
            return 1 + min_deletions_recursive(s, index + 1, open_count)


print(min_deletions_recursive("()())("))
print(min_deletions_recursive(")()("))
print(min_deletions_recursive("((()"))
```

Оба варианта дают одинаковый результат:

- `"()())("` -> нужно удалить 2 символа (лишнюю `)` на 5-й позиции и незакрытую `(` в конце);
- `")()("` -> нужно удалить 2 символа (первую `)` и последнюю `(`);
- `"((()"` -> нужно удалить 2 символа (две открывающие скобки остаются без пары).

## Задача № 7

Дана строка из круглых, квадратных и фигурных скобок. Нужно определить, сколько скобок минимально нужно **добавить**, чтобы вся последовательность стала правильной, а также желательно указать, какую именно скобку и в какую позицию нужно вставить.

Логика похожа на задачу № 6, только теперь работаем с тремя типами скобок и вместо удаления - считаем, сколько скобок не хватает, и сразу подбираем, какую скобку и куда нужно добавить.

Общая идея алгоритма:

1. Идем по строке, используя стек для открывающих скобок.
2. Если попалась закрывающая скобка:
   - если стек пуст или верхняя скобка в стеке другого типа - значит для этой закрывающей скобки не хватает открывающей. Запоминаем, что перед этим местом нужно добавить недостающую открывающую скобку соответствующего типа;
   - если типы совпадают, снимаем скобку со стека (пара найдена).
3. После окончания строки все скобки, оставшиеся в стеке, не получили пару - для каждой из них в конец строки нужно добавить закрывающую скобку того же типа.
4. Количество добавлений считается как сумма скобок, добавленных на шаге 2 и на шаге 3.

```python
def min_insertions(s):
    stack = []           # стек открывающих скобок с их позициями
    pairs_close = {'(': ')', '[': ']', '{': '}'}
    pairs_open = {')': '(', ']': '[', '}': '{'}

    to_insert = []        # список кортежей (позиция, что вставить)
    count = 0

    for i, ch in enumerate(s):
        if ch in '([{':
            stack.append((ch, i))
        elif ch in ')]}':
            if stack and stack[-1][0] == pairs_open[ch]:
                stack.pop()
            else:
                # для этой закрывающей скобки нет пары, значит перед ней
                # не хватает открывающей скобки того же типа
                to_insert.append((i, pairs_open[ch], "перед позицией"))
                count += 1

    # все, что осталось в стеке, нужно закрыть в конце строки
    for ch, pos in reversed(stack):
        to_insert.append((len(s), pairs_close[ch], "в конец строки"))
        count += 1

    return count, to_insert


def build_corrected_string(s):
    count, to_insert = min_insertions(s)

    # группируем скобки, которые нужно вставить, по позиции вставки
    insert_before = {}
    for pos, bracket, _ in to_insert:
        insert_before.setdefault(pos, []).append(bracket)

    # проходим по исходной строке и подставляем скобки в нужных местах,
    # порядок скобок в пределах одной позиции сохраняется
    result = []
    for i, ch in enumerate(s):
        result.extend(insert_before.get(i, []))
        result.append(ch)
    result.extend(insert_before.get(len(s), []))

    return count, "".join(result)


example = "{[(])}"
count, fixed = build_corrected_string(example)
print("Исходная строка:", example)
print("Нужно добавить скобок:", count)
print("Исправленная строка:", fixed)
```

На основе решения задачи № 7 составим собственный пример неправильной скобочной последовательности и приведем ее к правильному виду.

```python
my_sequence = "([{}]))[("
count, fixed = build_corrected_string(my_sequence)

print("Моя последовательность:", my_sequence)
print("Правильная ли она:", is_valid_brackets(my_sequence))
print("Сколько скобок нужно добавить:", count)
print("Исправленная последовательность:", fixed)
print("Проверка исправленной строки:", is_valid_brackets(fixed))
```

Разберем строку `"([{}]))[("` вручную:

- `(`, `[`, `{`, `}` - `{` закрылась корректно;
- `]` закрывает `[` - тоже корректно, стек после этого содержит только `(`;
- дальше идут `)` `)` - первая `)` закрывает `(`, стек становится пустым, а вторая `)` оказывается лишней, закрывать уже нечего;
- потом `[` и `(` открываются и остаются без пары до конца строки.

Значит не хватает открывающей скобки перед первой лишней `)`, а также нужно закрыть `[` и `(` в конце строки. Функция build_corrected_string подтверждает это и собирает корректную строку, которую можно проверить функцией is_valid_brackets из задачи № 5.
