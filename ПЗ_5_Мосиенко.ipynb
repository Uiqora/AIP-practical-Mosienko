```python
from collections import deque
from queue import Queue, LifoQueue

def make_queue(items):
    # собираем Queue из списка
    q = Queue()
    for x in items:
        q.put(x)
    return q


def queue_to_list(q):
    # Queue нельзя просто напечатать, поэтому достаем все элементы по порядку
    res = []
    while not q.empty():
        res.append(q.get())
    return res


def lifo_to_list(s):
    # LifoQueue достаем с вершины, потом разворачиваем (получается от дна к вершине)
    res = []
    while not s.empty():
        res.append(s.get())
    return res[::-1]
```

## Задача № 1

Дано число N (> 0) и набор из N чисел, создать стек из этих чисел (последнее число на вершине), извлечь все элементы, вывести их и количество извлеченных элементов

Стек реализовал тремя способами

Список: append и pop  
collections.deque: append и pop  
queue.LifoQueue: put и get

```python
data = [7, 12, 3, 20, 9]
N = len(data)
print('N =', N, 'числа:', data)

# 1) список
st = []
for x in data:
    st.append(x)
res = []
while st:
    res.append(st.pop())
print('list:', res, 'извлечено:', len(res))

# 2) deque
st = deque()
for x in data:
    st.append(x)
res = []
while st:
    res.append(st.pop())
print('deque:', res, 'извлечено:', len(res))

# 3) LifoQueue
st = LifoQueue()
for x in data:
    st.put(x)
res = []
while not st.empty():
    res.append(st.get())
print('LifoQueue:', res, 'извлечено:', len(res))
```

## Задача № 3

Дан непустой стек, создать два новых стека: в первый переместить все четные элементы исходного, во второй все нечетные (в новых стеках порядок обратный исходному, один из стеков может быть пустым)

Стек реализовал через deque и LifoQueue

Снимаем элементы с вершины исходного стека по одному  
Четное кладем в первый стек, нечетное во второй  
Так как снимаем сверху, порядок в новых стеках получается обратным

```python
data = [5, 8, 12, 3, 7, 10, 4]
print('Исходный стек:', data)

# deque
src = deque(data)
even = deque()
odd = deque()
while src:
    x = src.pop()
    if x % 2 == 0:
        even.append(x)
    else:
        odd.append(x)
print('deque, четные:', list(even))
print('deque, нечетные:', list(odd))

# LifoQueue
src = LifoQueue()
for x in data:
    src.put(x)
even = LifoQueue()
odd = LifoQueue()
while not src.empty():
    x = src.get()
    if x % 2 == 0:
        even.put(x)
    else:
        odd.put(x)
print('LifoQueue, четные:', lifo_to_list(even))
print('LifoQueue, нечетные:', lifo_to_list(odd))
```

## Задача № 4

Дан набор из N чисел (N > 10), создать очередь из этих чисел в том же порядке (первое число в начале, последнее в конце), извлечь все элементы, вывести их и количество извлеченных

Очередь реализовал тремя способами

Список: append и pop(0)  
collections.deque: append и popleft  
queue.Queue: put и get

```python
data = [11, 5, 8, 20, 3, 14, 9, 6, 17, 2, 13, 1]
N = len(data)
print('N =', N, 'числа:', data)

# 1) список
q = []
for x in data:
    q.append(x)
res = []
while q:
    res.append(q.pop(0))
print('list:', res, 'извлечено:', len(res))

# 2) deque
q = deque()
for x in data:
    q.append(x)
res = []
while q:
    res.append(q.popleft())
print('deque:', res, 'извлечено:', len(res))

# 3) Queue
q = make_queue(data)
res = queue_to_list(q)
print('Queue:', res, 'извлечено:', len(res))
```

## Задача № 8

Дано число N (> 0) и непустая очередь, создать функцию, которая извлекает из очереди N начальных элементов и выводит их (если элементов меньше N, извлекаем все)

В цикле на N шагов извлекаем элемент из начала  
Если очередь опустела раньше, выходим из цикла

```python
def take_first(q, n):
    res = []
    for i in range(n):
        if not q:  # очередь кончилась
            break
        res.append(q.popleft())
    print('Извлечено:', res)


q = deque([5, 10, 15, 20, 25, 30])
print('Очередь:', list(q))
take_first(q, 4)
print('Осталось:', list(q))

print()
take_first(q, 10)  # N больше размера очереди
print('Осталось:', list(q))
```

## Задача № 9

Дана непустая очередь, извлекать элементы, пока начальный элемент не станет четным, и выводить извлеченные значения (если четных нет, извлекаем все)

Извлекаем элемент, если он нечетный, запоминаем и идем дальше  
Если встретили четный, возвращаем его обратно в начало (appendleft) и останавливаемся, так как он должен остаться начальным

```python
def pop_until_even(q):
    res = []
    while q:
        x = q.popleft()
        if x % 2 == 0:
            q.appendleft(x)  # четный остается в начале очереди
            break
        res.append(x)
    return res


for items in ([3, 7, 9, 4, 1], [5, 11, 13], [2, 3, 5]):
    q = deque(items)
    print('Очередь:', items)
    print('Извлечено:', pop_until_even(q), 'осталось:', list(q))
```

## Задача № 11

Дано число N (> 0) и две непустые очереди, создать функцию, которая перемещает N начальных элементов первой очереди в конец второй (если в первой меньше N элементов, перемещаем все)

Очереди реализовал через deque и Queue, для каждого типа своя функция

```python
def move_deque(q1, q2, n):
    for i in range(n):
        if not q1:  # первая очередь кончилась
            break
        q2.append(q1.popleft())


def move_queue(q1, q2, n):
    for i in range(n):
        if q1.empty():
            break
        q2.put(q1.get())


a = [3, 6, 9, 12, 15, 18]
b = [1, 2]

for n in (4, 10):
    print('N =', n)

    q1, q2 = deque(a), deque(b)
    move_deque(q1, q2, n)
    print('deque:', list(q1), list(q2))

    q1, q2 = make_queue(a), make_queue(b)
    move_queue(q1, q2, n)
    print('Queue:', queue_to_list(q1), queue_to_list(q2))
```

## Задача № 13

Даны две непустые очереди, элементы каждой упорядочены по возрастанию, объединить их в одну с сохранением упорядоченности

Держим по одному текущему элементу из каждой очереди (a и b)  
Меньший отправляем в результат и берем следующий из той же очереди  
Когда одна очередь закончилась, остаток другой дописываем в конец, он уже упорядочен  
Смотрим только на элементы, извлеченные из начала, к середине очереди не обращаемся

Очереди реализовал через deque и Queue

```python
def merge_deque(q1, q2):
    res = deque()
    a = q1.popleft()
    b = q2.popleft()
    while True:
        if a <= b:
            res.append(a)
            if not q1:  # первая закончилась, дописываем b и остаток второй
                res.append(b)
                while q2:
                    res.append(q2.popleft())
                break
            a = q1.popleft()
        else:
            res.append(b)
            if not q2:  # вторая закончилась, дописываем a и остаток первой
                res.append(a)
                while q1:
                    res.append(q1.popleft())
                break
            b = q2.popleft()
    return res


def merge_queue(q1, q2):
    res = Queue()
    a = q1.get()
    b = q2.get()
    while True:
        if a <= b:
            res.put(a)
            if q1.empty():
                res.put(b)
                while not q2.empty():
                    res.put(q2.get())
                break
            a = q1.get()
        else:
            res.put(b)
            if q2.empty():
                res.put(a)
                while not q1.empty():
                    res.put(q1.get())
                break
            b = q2.get()
    return res


tests = [
    ([1, 4, 6, 10], [2, 3, 8, 12, 15]),
    ([5, 6], [1, 2, 3]),
    ([2, 2, 7], [2, 9]),
]

for a, b in tests:
    r1 = list(merge_deque(deque(a), deque(b)))
    r2 = queue_to_list(merge_queue(make_queue(a), make_queue(b)))
    print(a, b)
    print('deque:', r1, r1 == sorted(a + b))
    print('Queue:', r2, r2 == sorted(a + b))
```
